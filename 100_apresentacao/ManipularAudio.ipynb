```python
import numpy as np
from IPython.display import Audio

taxaAmostragem = 16000
bits = 16
```

**Audio Seno de 1 kHz gerado no ocenaudio**

```python
Seno = np.fromfile("Seno_1K.pcm", dtype=np.int16)
Audio(Seno,rate=taxaAmostragem)
```

**Voz gravada**

```python
Voz = np.fromfile("Voz.pcm", dtype=np.int16)
Voz = Voz[:taxaAmostragem*10]
Audio(Voz,rate=taxaAmostragem)
```

**Metade do Seno de 1 Khz**

```python
MetadeSeno = Seno/2
Audio(MetadeSeno,rate=taxaAmostragem)
MetadeSeno.tofile("MetadeSeno.pcm")
```

```python
mix = Voz.astype(np.int32) + Seno.astype(np.int32)
mix = np.clip(mix,-32768,32767)
mix = mix.astype(np.int16) 
Audio(mix,rate=taxaAmostragem)
mix.tofile("mixagem.pcm")
```
