```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import style
```

# Amostra um sinal pseudo continuo e apresenta o sinal amostrado #

```python
# Sinal analógico

Dt = 5e-5

t = np.arange(
    -5e-3,
    5e-3 + Dt,
    Dt
)

f = 1000

xa = np.cos(2 * np.pi * f * t)

# Sinal discreto

Fs = 200
Ts = 1 / Fs

n = np.linspace(-20, 21)

xd = np.cos(2 * np.pi * n * f / Fs)

# Figura

fig, ax = plt.subplots(2, 1, figsize=(14, 9))

# Primeiro gráfico

ax[0].plot(
    t * 1000,
    xa,
    label='Sinal analógico'
)
ax[0].stem(
    n * Ts * 1000,
    xd,
    linefmt='r-',
    markerfmt='ro',
    basefmt='r-',
    label='Amostras'
)
ax[0].set_xlim(-5, 5)
ax[0].set_ylim(-1.2, 1.2)
ax[0].set_xlabel('Tempo [ms]')
ax[0].set_ylabel('xa(t)')
ax[0].set_title(
    f'Sinal analógico e amostragem — f = {f} Hz, Fs = {Fs} Hz'
)
ax[0].grid(True)
ax[0].legend()

# Segundo gráfico

ax[1].stem(
    n,
    xd,
    linefmt='r-',
    markerfmt='ro',
    basefmt='r-'
)
ax[1].set_xlim(-20, 20)
ax[1].set_ylim(-1.2, 1.2)
ax[1].set_xlabel('Amostra n')
ax[1].set_ylabel('x[n]')
ax[1].set_title('Sinal amostrado')
ax[1].grid(True)


plt.tight_layout()
plt.show()
```

# Impulso Unitário #

```python
x = []
n = []

for i in range(-1,6):
    n.append(i)
    if i == 0:
        x.append(1)
    else:
        x.append(0)

plt.figure(figsize= (18,8)) #defini o tamanho da figura
style.use('dark_background')
plt.rcParams['xtick.labelsize'] = 15
plt.rcParams['ytick.labelsize'] = 15
plt.stem(n,x,'yo', label = 'Função de Impulso Unitário')
plt.ylim(-0.01,1.2)
plt.xlim(-1,7)
plt.xlabel('Tempo', fontsize = 15)
plt.ylabel('Amplitude', fontsize = 15)
plt.legend(fontsize = 20)
plt.show()
```

# Degrau Unitário #

```python
x = []
n = []

for i in range(-1,8):
    n.append(i)
    if i >= 0:
        x.append(1)
    else:
        x.append(0)

plt.figure(figsize= (18,8)) #defini o tamanho da figura
style.use('dark_background')
plt.rcParams['xtick.labelsize'] = 15
plt.rcParams['ytick.labelsize'] = 15
plt.stem(n,x,'yo', label = 'Função de Degrau Unitário')
plt.ylim(-0.01,1.2)
plt.xlim(-3,7)
plt.xlabel('Tempo', fontsize = 15)
plt.ylabel('Amplitude', fontsize = 15)
plt.legend(fontsize = 20)
plt.show()
```

# Função Sinusoidal #

```python
n = np.linspace(0,10,100)
amp = 1
f = 100
x = amp * np.sin(2 * np.pi * f * n)

plt.figure(figsize= (18,12)) #defini o tamanho da figura
style.use('dark_background')
plt.rcParams['xtick.labelsize'] = 15
plt.rcParams['ytick.labelsize'] = 15
plt.stem(n,x,'yo', label = 'Função Sinusoidal')
plt.ylim(-1,1)
plt.xlim(0,10)
plt.xlabel('Tempo', fontsize = 15)
plt.ylabel('Amplitude', fontsize = 15)
plt.legend(fontsize = 15)
plt.show()
```

# Função Exponencial #

```python
n = np.linspace(0,10,100)
amp = 1
x = amp * np.exp(-n)

plt.figure(figsize= (18,12)) #defini o tamanho da figura
style.use('dark_background')
plt.rcParams['xtick.labelsize'] = 15
plt.rcParams['ytick.labelsize'] = 15
plt.stem(n*10,x,'yo', label = 'Função Exponencial')
plt.ylim(0,1)
plt.xlim(0,30)
plt.xlabel('Tempo', fontsize = 15)
plt.ylabel('Amplitude', fontsize = 15)
plt.legend(fontsize = 15)
plt.show()
```
